```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set(style="whitegrid")

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("retail_sales_dataset.csv")
df.head()
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"])

# Verify the conversion
print(df.dtypes)
```

```python
print("Descriptive Statistics")
display(df.describe())

print("\nMean")
print(df.mean(numeric_only=True))

print("\nMedian")
print(df.median(numeric_only=True))

print("\nMode")
display(df.mode())

print("\nStandard Deviation")
print(df.std(numeric_only=True))
```

## Observation

- The dataset contains 1,000 retail transactions.
- There are no missing values.
- There are no duplicate records.
- The Date column has been converted to datetime format for time-series analysis.

```python
# Create Month column
df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Month")["Total Amount"].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot(marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

## Observation

- The line chart shows how total sales change month by month.
- Some months generate higher revenue than others, indicating seasonal demand.
- This analysis helps businesses plan inventory and marketing campaigns during peak periods.

```python
# Create Quarter column
df["Quarter"] = df["Date"].dt.to_period("Q").astype(str)

quarterly_sales = df.groupby("Quarter")["Total Amount"].sum()

plt.figure(figsize=(8,5))
quarterly_sales.plot(marker="o")
plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.show()
```

## Observation

- The quarterly sales trend provides a broader view of business performance.
- Comparing quarterly revenue helps identify long-term growth patterns and seasonal fluctuations.
- This information can support strategic planning and budgeting decisions.

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Age", bins=10, kde=True)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

## Observation

The age distribution highlights the customer age ranges represented in the dataset and helps identify potential target customer groups.

```python
plt.figure(figsize=(6,5))
sns.countplot(data=df, x="Gender")
plt.title("Gender Distribution of Customers")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

## Observation

The gender distribution shows the composition of the customer base and can support targeted marketing strategies.

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

## Observation

Categories with the highest revenue contribute most strongly to overall business performance and should receive appropriate inventory and promotional attention.

```python
top_quantity = df.groupby("Product Category")["Quantity"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
sns.barplot(x=top_quantity.index, y=top_quantity.values)
plt.title("Top Selling Product Categories")
plt.xlabel("Product Category")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.show()
```

## Observation

High sales volume does not necessarily mean the highest revenue. Comparing quantity and revenue helps distinguish high-volume categories from high-value categories.

```python
plt.figure(figsize=(8,6))
correlation = df.corr(numeric_only=True)
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## Observation

The correlation heatmap shows the relationships between numerical variables and helps identify variables that move together.

```python
df["Age Group"] = pd.cut(df["Age"], bins=[18,25,35,45,55,65], labels=["18-25","26-35","36-45","46-55","56-65"])
plt.figure(figsize=(10,6))
sns.barplot(data=df, x="Age Group", y="Total Amount", hue="Gender")
plt.title("Average Spending by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Average Transaction Amount")
plt.show()
```

## Observation

This visualization compares average transaction amounts across age groups and genders, helping identify customer segments with stronger spending behavior.
