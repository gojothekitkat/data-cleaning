# Data Cleaning on Titanic Dataset

## Objective
Systematically transform a messy dataset into a clean, analysis-ready dataset while documenting data-quality decisions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
sns.set(style='whitegrid')
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

```python
before_rows = len(df)
before_cols = len(df.columns)
before_nulls = int(df.isnull().sum().sum())
duplicates_before = int(df.duplicated().sum())
print('Shape:', df.shape)
print('\nData Types:')
print(df.dtypes)
print('\nMissing Values:')
print(df.isnull().sum())
print('\nDuplicate Rows:', duplicates_before)
display(df.describe())
```

## Observation
The initial quality report identifies missing values, particularly in Age, Cabin and Embarked. The dataset also contains a mix of numerical and categorical variables that require different cleaning strategies.

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])
print(df.isnull().sum())
```

## Missing Data Decisions
- **Age:** median imputation was selected because it is robust to outliers.
- **Embarked:** mode imputation was selected because it is categorical.
- **Cabin:** the column was removed because most observations were missing, making reliable imputation difficult.

```python
duplicates_before = df.duplicated().sum()
df = df.drop_duplicates()
duplicates_after = df.duplicated().sum()
print('Duplicates Before:', duplicates_before)
print('Duplicates After:', duplicates_after)
```

## Observation
The dataset was checked for duplicate records. No duplicate rows were found, so no records were removed during this step.

```python
df['Sex'] = df['Sex'].str.capitalize()
df['Embarked'] = df['Embarked'].replace({'S':'Southampton','C':'Cherbourg','Q':'Queenstown'})
df[['Sex','Embarked']].head()
```

## Observation
Categorical values were standardized for readability and consistency. Sex values use consistent capitalization and port codes were expanded to their full names.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Fare'])
plt.title('Boxplot of Fare')
plt.show()
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['Fare'] < lower) | (df['Fare'] > upper)]
print('Number of Fare Outliers:', len(outliers))
```

## Observation
The IQR method identifies unusually high Fare values. These observations are retained because high fares can represent genuine passenger classes or travel arrangements rather than data-entry errors.

```python
print(df.dtypes)
```

## Observation
The final data types were reviewed to ensure numerical and categorical fields remain appropriate for analysis.

```python
summary = pd.DataFrame({
    'Metric':['Rows','Columns','Missing Values','Duplicate Rows'],
    'Before Cleaning':[before_rows,before_cols,before_nulls,0],
    'After Cleaning':[len(df),len(df.columns),int(df.isnull().sum().sum()),int(df.duplicated().sum())]
})
display(summary)
```

## Observation
The before-versus-after summary demonstrates the impact of the cleaning process: missing values were addressed, the heavily incomplete Cabin column was removed, and duplicate checks were completed.

```python
df.to_csv('Titanic_Cleaned.csv', index=False)
print('Cleaned dataset saved successfully as Titanic_Cleaned.csv')
```

# Conclusion

The Titanic dataset was transformed into a cleaner, analysis-ready dataset by handling missing values, removing an excessively incomplete column, standardizing categorical values, checking duplicates, detecting Fare outliers with the IQR method, and verifying data types. The cleaned CSV can be used for further exploratory analysis or modelling.
